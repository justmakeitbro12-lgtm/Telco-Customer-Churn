```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

categorical_cols = df1[['gender', 'Partner', 'Dependents','Dependents', 'MultipleLines', "InternetService", 'OnlineSecurity', 'OnlineBackup', 'DeviceProtection','TechSupport','StreamingTV','StreamingMovies','Contract','PaperlessBilling','PaymentMethod','Churn']]

for col in categorical_cols:
    df1[col] = le.fit_transform(df1[col])

categorical_cols
```

```python
from sklearn.preprocessing import StandardScaler

# Define column names as a list, not a DataFrame
numerical_cols = ['SeniorCitizen', 'tenure', 'MonthlyCharges', 'TotalCharges']

scaler = StandardScaler()

# Apply scaling to the specified columns
df1[numerical_cols] = scaler.fit_transform(df1[numerical_cols])

numerical_cols
```
